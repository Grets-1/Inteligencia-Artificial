# Algoritmo  de Distancia de Levenshtein es un  

```python
def levenshtein_futbol(palabra_a: str, palabra_b: str) -> int:
    """Calcula la distancia de edición entre dos palabras.

    Args:
        palabra_a (str): Primera cadena. Ej: ``"Gool"``.
        palabra_b (str): Segunda cadena. Ej: ``"Gol"``.

    Returns:
        int: Distancia mínima de edición (valor ≥ 0).

    Raises:
        TypeError: Si alguno de los argumentos no es ``str``.

    Examples:
        >>> levenshtein_futbol("Messi", "Messi")
        0
        >>> levenshtein_futbol("Gol", "Gool")
        1
    """
    try:
        if not isinstance(palabra_a, str) or not isinstance(
            palabra_b, str
        ):
            raise TypeError(
                "Ambos argumentos deben ser str. "
                f"Recibido: {type(palabra_a)}, {type(palabra_b)}."
            )

        n: int = len(palabra_a)
        m: int = len(palabra_b)

        # 1. Crear la matriz inicializada en ceros  (n+1) × (m+1)
        matriz: List[List[int]] = [
            [0 for _ in range(m + 1)] for _ in range(n + 1)
        ]

        # 2. Casos base: transformar cadena a vacío o viceversa
        for i in range(n + 1):
            matriz[i][0] = i      # costo = i eliminaciones
        for j in range(m + 1):
            matriz[0][j] = j      # costo = j inserciones

        # 3. Llenado de la tabla con programación dinámica
        for i in range(1, n + 1):
            for j in range(1, m + 1):
                # Costo 0 si caracteres son iguales, 1 si difieren
                if palabra_a[i - 1] == palabra_b[j - 1]:
                    costo: int = 0
                else:
                    costo = 1

                # Mínimo entre las 3 operaciones posibles
                matriz[i][j] = min(
                    matriz[i - 1][j] + 1,       # Eliminación
                    matriz[i][j - 1] + 1,       # Inserción
                    matriz[i - 1][j - 1] + costo  # Sustitución
                )

        distancia: int = matriz[n][m]
        logger.info(
            "Levenshtein('%s', '%s') = %d",
            palabra_a, palabra_b, distancia,
        )
        return distancia

    except TypeError as exc:
        logger.error("TypeError en levenshtein_futbol(): %s", exc)
        raise

# --- PRUEBA PARA CLASE ---
termino_busqueda = "Gool"
termino_real = "Gol"

distancia = levenshtein_futbol(termino_busqueda, termino_real)
print(f"La distancia entre '{termino_busqueda}' y '{termino_real}' es: {distancia}")
# Resultado esperado: 1
```

# Word2Vec Algoritmo desde Cero

```python
import numpy as np

def softmax(x: np.ndarray) -> np.ndarray:
    """
    Aplica Softmax estable numéricamente a un vector 1-D.

    Args:
        x (np.ndarray): Vector de logits, shape ``(V,)``.

    Returns:
        np.ndarray: Distribución de probabilidad, shape ``(V,)``,
            con valores en (0, 1) que suman exactamente 1.
    """
    try:
        if x.ndim != 1:
            raise ValueError(
                f"Se esperaba array 1-D; shape recibido: {x.shape}."
            )
        # Estabilidad: restar el máximo antes de exponenciar
        e_x: np.ndarray = np.exp(x - np.max(x))
        return e_x / e_x.sum(axis=0)

    except ValueError as exc:
        logger.error("ValueError en softmax(): %s", exc)
        raise


class Word2VecSimplified:
    """
    Implementación manual de Word2Vec (Skip-Gram, 1 capa oculta).


    Args:
        vocab_size (int): Tamaño del vocabulario V.
        embed_dist (int): Dimensión del espacio de embedding N.

    Attributes:
        w1 (np.ndarray): Pesos de entrada, shape ``(V, N)``.
        w2 (np.ndarray): Pesos de salida,  shape ``(N, V)``.
    """

    def __init__(self, vocab_size: int, embed_dist: int) -> None:
        """Inicializa pesos con distribución normal estándar.

        Args:
            vocab_size (int): Número de palabras en el vocabulario (V).
            embed_dist (int): Dimensión del embedding (N).
        """
        try:
            if vocab_size <= 0 or embed_dist <= 0:
                raise ValueError(
                    "vocab_size y embed_dist deben ser enteros > 0. "
                    f"Recibido: vocab_size={vocab_size}, "
                    f"embed_dist={embed_dist}."
                )

            self.n: int = embed_dist   # N: dimensión del embedding
            self.v: int = vocab_size   # V: tamaño del vocabulario

            # Pesos de entrada W1  shape (V × N)
            self.w1: np.ndarray = np.random.randn(self.v, self.n)
            # Pesos de salida  W2  shape (N × V)
            self.w2: np.ndarray = np.random.randn(self.n, self.v)

            logger.info(
                "Word2VecSimplified – V=%d, N=%d", vocab_size, embed_dist
            )

        except ValueError as exc:
            logger.error(
                "ValueError en Word2VecSimplified.__init__(): %s", exc
            )
            raise

    def forward(
        self, x: int
    ) -> Tuple[np.ndarray, np.ndarray, np.ndarray]:
        """Propagación hacia adelante para un índice de palabra.

        Args:
            x (int): Índice del token de entrada (0 ≤ x < V).

        Returns:
            Tuple[np.ndarray, np.ndarray, np.ndarray]:
                * ``y_c`` – Probabilidades softmax, shape ``(V,)``.
                * ``h``   – Capa oculta (embedding), shape ``(N,)``.
                * ``u``   – Logits antes de softmax, shape ``(V,)``.

        Raises:
            IndexError: Si ``x`` está fuera del rango [0, V-1].
        """
        try:
            if not (0 <= x < self.v):
                raise IndexError(
                    f"Índice x={x} fuera de rango [0, {self.v - 1}]."
                )

            # h = W1[x]  → fila x de la matriz de pesos (embedding)
            h: np.ndarray = self.w1[x]

            # u = h · W2  → proyección al espacio del vocabulario
            u: np.ndarray = np.dot(h, self.w2)

            # ŷ = softmax(u)  → distribución de probabilidad
            y_c: np.ndarray = softmax(u)

            return y_c, h, u

        except (IndexError, ValueError) as exc:
            logger.error("Error en forward(): %s", exc)
            raise

    def backward(
        self, e: np.ndarray, h: np.ndarray, x: int
    ) -> None:
        """Retropropagación y actualización de pesos (lr = 0.01).

        Args:
            e (np.ndarray): Error = ŷ − y_true, shape ``(V,)``.
            h (np.ndarray): Capa oculta del forward, shape ``(N,)``.
            x (int):        Índice del token de entrada.

        Raises:
            ValueError: Si shapes de ``e`` o ``h`` son incorrectos.
        """
        try:
            if e.shape != (self.v,):
                raise ValueError(
                    f"e debe tener shape ({self.v},); "
                    f"recibido: {e.shape}."
                )
            if h.shape != (self.n,):
                raise ValueError(
                    f"h debe tener shape ({self.n},); "
                    f"recibido: {h.shape}."
                )

            # Gradiente de W2: producto externo h ⊗ e  → shape (N×V)
            dw2: np.ndarray = np.outer(h, e)

            # Gradiente hacia W1: W2 · e  → shape (N,)
            dw1: np.ndarray = np.dot(self.w2, e.T)

            # Actualización con tasa de aprendizaje lr = 0.01
            self.w2 = self.w2 - (0.01 * dw2)
            self.w1[x] = self.w1[x] - (0.01 * dw1.T)

            logger.debug(
                "backward() OK – ||dw2||=%.4f ||dw1||=%.4f",
                float(np.linalg.norm(dw2)),
                float(np.linalg.norm(dw1)),
            )

        except (ValueError, IndexError) as exc:
            logger.error("Error en backward(): %s", exc)
            raise

# Ejemplo de uso conceptual:
# 1. Definimos vocabulario: {'gol': 0, 'messi': 1, 'futbol': 2}
# 2. Si entra 'messi' (index 1) y el contexto es 'gol' (index 0)
modelo = Word2VecSimplified(vocab_size=3, embed_dist=2)
y_pred, h, u = modelo.forward(1)

# El error es (predicción - real) donde real es un one-hot del contexto
error = y_pred.copy()
error[0] -= 1

modelo.backward(error, h, 1)
print("Vector actualizado de 'Messi':", modelo.w1[1])
```

# Sobel

```python
import numpy as np

def sobel_filter(image: np.ndarray) -> np.ndarray:
    """Aplica el operador de Sobel para detectar bordes en una imagen.

    Kernels de Sobel
    ----------------
    Gradiente horizontal Gx (detecta bordes verticales):

        Gx = [[-1,  0,  1],
               [-2,  0,  2],
               [-1,  0,  1]]

    Gradiente vertical Gy (detecta bordes horizontales):

        Gy = [[ 1,  2,  1],
               [ 0,  0,  0],
               [-1, -2, -1]]

    Magnitud del gradiente en cada píxel (i, j):

        G(i,j) = sqrt( Gx(i,j)² + Gy(i,j)² )

    Normalización a rango [0, 255]:

        G_norm = ( G / G_max ) * 255

    Args:
        image (np.ndarray): Imagen 2-D en escala de grises
            (dtype float o uint8).

    Returns:
        np.ndarray: Mapa de bordes normalizado a ``uint8``
            con valores en [0, 255].

    Raises:
        TypeError : Si ``image`` no es ``np.ndarray``.
        ValueError: Si ``image`` no es 2-D o es menor a 3×3.

    Notes:
        Se usa padding de ceros (``mode='constant'``) para conservar
        las dimensiones originales de la imagen de entrada.
    """
    try:
        if not isinstance(image, np.ndarray):
            raise TypeError(
                "image debe ser np.ndarray; "
                f"recibido: {type(image).__name__}."
            )
        if image.ndim != 2:
            raise ValueError(
                "image debe ser 2-D (escala de grises); "
                f"ndim recibido: {image.ndim}."
            )
        if image.shape[0] < 3 or image.shape[1] < 3:
            raise ValueError(
                "La imagen debe tener al menos 3×3 píxeles."
            )

        # ── Kernels de Sobel ──────────────────────────────────────
        # Gx: responde a cambios horizontales de intensidad
        gx_kernel: np.ndarray = np.array(
            [[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]
        )
        # Gy: responde a cambios verticales de intensidad
        gy_kernel: np.ndarray = np.array(
            [[1, 2, 1], [0, 0, 0], [-1, -2, -1]]
        )

        rows, cols = image.shape
        output: np.ndarray = np.zeros((rows, cols))

        # Padding de ceros para no perder los bordes de la imagen
        padded_img: np.ndarray = np.pad(
            image, (1, 1), mode="constant"
        )

        # Convolución manual píxel a píxel con ventana 3×3
        for i in range(rows):
            for j in range(cols):
                region = padded_img[i: i + 3, j: j + 3]

                # Suma de Hadamard (elemento a elemento) × kernel
                gx: float = float(np.sum(region * gx_kernel))
                gy: float = float(np.sum(region * gy_kernel))

                # Magnitud euclidiana del gradiente
                output[i, j] = np.sqrt(gx ** 2 + gy ** 2)

        # Normalización al rango [0, 255]
        output = (output / output.max()) * 255

        logger.info(
            "Sobel aplicado – shape entrada: %s", image.shape
        )
        return output.astype(np.uint8)

    except (TypeError, ValueError) as exc:
        logger.error("Error en sobel_filter(): %s", exc)
        raise

    # Normalización a 0-255
    output = (output / output.max()) * 255
    return output.astype(np.uint8)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, color

# 1. Cargar una imagen de muestra (ej. astronauta) y convertir a gris
image = color.rgb2gray(data.astronaut())

# 2. Aplicar la función sobel_filter que definimos anteriormente
# (Asegúrate de que la función sobel_filter esté definida en el mismo script)
bordes = sobel_filter(image)

# 3. Visualización comparativa
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title("Imagen Original")
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Resultado: Operador de Sobel")
plt.imshow(bordes, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from skimage import data, color, img_as_ubyte # Importamos esta utilidad

def extraer_caracteristicas_sift(
    imagen_uint8: np.ndarray,
) -> Tuple[List[cv2.KeyPoint], Optional[np.ndarray]]:
    """Extrae puntos clave y descriptores SIFT de una imagen.

    SIFT (Scale-Invariant Feature Transform) detecta puntos de interés
    que son invariantes a escala, rotación e iluminación parcial.

    Pasos internos de SIFT (ejecutados por OpenCV)
    -----------------------------------------------
    1. **Espacio-escala / DoG**: Se construye una pirámide de imágenes
       desenfocadas con distintos σ y se calcula la Diferencia de
       Gaussianas (DoG = L(x,y,kσ) − L(x,y,σ)) para encontrar
       extremos locales candidatos a puntos clave.

    2. **Localización de puntos clave**: Se refina la posición sub-píxel
       mediante la expansión de Taylor de la función DoG y se descartan
       puntos de bajo contraste y sobre bordes (usando la razón de
       curvaturas principales con la matriz Hessiana).

    3. **Asignación de orientación**: Se calcula el histograma de
       gradientes en la vecindad del punto; la orientación dominante
       hace al descriptor invariante a rotación.

    4. **Generación del descriptor**: En una ventana 16×16 centrada en
       el punto clave se computan 8 histogramas de gradientes de 4×4
       celdas → vector de 128 dimensiones normalizado.

    Args:
        imagen_uint8 (np.ndarray): Imagen en escala de grises de tipo
            ``uint8`` (valores en [0, 255]), shape ``(H, W)``.

    Returns:
        Tuple[List[cv2.KeyPoint], Optional[np.ndarray]]:
            * ``kp``  – Lista de puntos clave detectados.
            * ``des`` – Matriz de descriptores, shape ``(N, 128)``
              donde N = número de puntos clave. ``None`` si no hay.

    Raises:
        TypeError : Si ``imagen_uint8`` no es ``np.ndarray``.
        ValueError: Si la imagen no es 2-D o no es de tipo uint8.

    Notes:
        La imagen **debe** estar en uint8 (0-255). Si se parte de una
        imagen float (0.0-1.0) usar ``skimage.img_as_ubyte()``
        antes de llamar a esta función.
    """
    try:
        if not isinstance(imagen_uint8, np.ndarray):
            raise TypeError(
                "imagen_uint8 debe ser np.ndarray; "
                f"recibido: {type(imagen_uint8).__name__}."
            )
        if imagen_uint8.ndim != 2:
            raise ValueError(
                "La imagen debe ser 2-D (escala de grises); "
                f"ndim recibido: {imagen_uint8.ndim}."
            )
        if imagen_uint8.dtype != np.uint8:
            raise ValueError(
                "La imagen debe ser de tipo uint8. "
                f"Recibido: {imagen_uint8.dtype}. "
                "Usa img_as_ubyte() para convertir desde float."
            )

        # Crear el detector SIFT de OpenCV
        sift: cv2.SIFT = cv2.SIFT_create()

        # detectAndCompute devuelve puntos clave y descriptores 128-D
        kp: List[cv2.KeyPoint]
        des: Optional[np.ndarray]
        kp, des = sift.detectAndCompute(imagen_uint8, None)

        logger.info(
            "SIFT – puntos clave: %d | "
            "descriptor shape: %s",
            len(kp),
            des.shape if des is not None else "None",
        )
        return kp, des

    except (TypeError, ValueError) as exc:
        logger.error("Error en extraer_caracteristicas_sift(): %s", exc)
        raise
    except cv2.error as exc:
        logger.error("Error de OpenCV en SIFT: %s", exc)
        raise

# 1. Cargar imagen
image_float = color.rgb2gray(data.astronaut())

# 2. CORRECCIÓN: Convertir de float (0.0-1.0) a uint8 (0-255)
image_uint8 = img_as_ubyte(image_float)

# 3. Ahora sí, ejecutar
kp, des = extraer_caracteristicas_sift(image_uint8)

print(f"Puntos clave encontrados: {len(kp)}")
print(f"Dimensión del descriptor: {des.shape}") # Nota: des es una matriz completa
```

```python
def gaussian_kernel(size, sigma):
    ax = np.arange(-(size // 2), size // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)
    return kernel
```

```python
kernel = gaussian_kernel(5, 1)

print(kernel)
```

```python
def convolution(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    pad_h = kh // 2
    pad_w = kw // 2
    padded = np.pad(
        image,
        ((pad_h, pad_h), (pad_w, pad_w)),
        mode='constant'
    )
    output = np.zeros_like(image, dtype=np.float32)
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)
    return output
```

```python
def detect_keypoints(dog, threshold=5):
    keypoints = []
    h, w = dog.shape
    for i in range(1, h-1):
        for j in range(1, w-1):
            patch = dog[i-1:i+2, j-1:j+2]
            center = dog[i, j]
            if center == np.max(patch) or center == np.min(patch):
                if abs(center) > threshold:
                    keypoints.append((j, i))
    return keypoints
```

```python
def compute_gradients(image):

    image = image.astype(np.float32)

    h, w = image.shape

    magnitude = np.zeros((h, w), dtype=np.float32)
    orientation = np.zeros((h, w), dtype=np.float32)

    for i in range(1, h-1):
        for j in range(1, w-1):

            # derivadas
            dx = image[i, j+1] - image[i, j-1]
            dy = image[i+1, j] - image[i-1, j]

            # magnitud
            mag = np.sqrt(dx**2 + dy**2)

            # orientación
            angle = np.arctan2(dy, dx)

            magnitude[i, j] = mag
            orientation[i, j] = angle

    return magnitude, orientation
```

```python
def orientation_histogram(
    magnitude,
    orientation,
    x,
    y,
    radius=8,
    bins=36
):

    hist = np.zeros(bins)
    h, w = magnitude.shape
    for i in range(-radius, radius):
        for j in range(-radius, radius):
            yi = y + i
            xj = x + j
            if yi < 0 or yi >= h:
                continue
            if xj < 0 or xj >= w:
                continue
            mag = magnitude[yi, xj]
            angle = orientation[yi, xj]
            angle_deg = np.degrees(angle) % 360
            bin_idx = int(angle_deg / (360 / bins))
            hist[bin_idx] += mag

    return hist
```

```python
def simple_descriptor(
    magnitude,
    orientation,
    x,
    y,
    window=16,
    bins=8
):

    descriptor = []

    step = window // 4

    for i in range(4):
        for j in range(4):
            hist = np.zeros(bins)
            for wi in range(step):
                for wj in range(step):
                    yi = y + i*step + wi
                    xj = x + j*step + wj
                    if yi >= magnitude.shape[0]:
                        continue
                    if xj >= magnitude.shape[1]:
                        continue
                    mag = magnitude[yi, xj]
                    angle = orientation[yi, xj]
                    angle_deg = np.degrees(angle) % 360
                    bin_idx = int(angle_deg / (360 / bins))
                    hist[bin_idx] += mag
            descriptor.extend(hist)
    descriptor = np.array(descriptor)

    # normalización
    descriptor = descriptor / (np.linalg.norm(descriptor) + 1e-7)

    return descriptor
```

```python
def generate_descriptors(
    magnitude,
    orientation,
    keypoints
):

    descriptors = []
    valid_keypoints = []
    for x, y in keypoints:

        # evitar bordes
        if y + 16 >= magnitude.shape[0]:
            continue
        if x + 16 >= magnitude.shape[1]:
            continue
        descriptor = simple_descriptor(
            magnitude,
            orientation,
            x,
            y
        )
        descriptors.append(descriptor)
        valid_keypoints.append((x, y))
    return np.array(descriptors), valid_keypoints
```

```python
def euclidean_distance(a, b):

    return np.linalg.norm(a - b)
```

```python
def match_descriptors(
    desc1,
    desc2,
    threshold=0.5
):
    matches = []
    for i, d1 in enumerate(desc1):
        best_distance = float("inf")
        best_j = -1
        for j, d2 in enumerate(desc2):
            dist = euclidean_distance(d1, d2)
            if dist < best_distance:
                best_distance = dist
                best_j = j
        if best_distance < threshold:
            matches.append((i, best_j, best_distance))
    return matches
```

```python
def draw_matches(
    img1,
    img2,
    kp1,
    kp2,
    matches
):

    h1, w1 = img1.shape
    h2, w2 = img2.shape

    canvas = np.zeros(
        (max(h1, h2), w1 + w2, 3),
        dtype=np.uint8
    )

    canvas[:h1, :w1] = cv2.cvtColor(img1, cv2.COLOR_GRAY2BGR)
    canvas[:h2, w1:] = cv2.cvtColor(img2, cv2.COLOR_GRAY2BGR)
    for i, j, dist in matches:
        x1, y1 = kp1[i]
        x2, y2 = kp2[j]
        color = (0,255,0)
        cv2.circle(canvas, (x1, y1), 3, color, -1)
        cv2.circle(canvas, (x2+w1, y2), 3, color, -1)
        cv2.line(
            canvas,
            (x1, y1),
            (x2+w1, y2),
            color,
            1
        )

    return canvas
```

# Detección y Mitigación de Sesgo Algorítmico

**Objetivos de Aprendizaje**

- Identificar y medir el sesgo en un conjunto de datos utilizando métricas de equidad (fairness metrics).

- Entrenar un clasificador base y evaluar su impacto dispar en un atributo protegido.

- Aplicar técnicas algorítmicas de mitigación (post-procesamiento o reducción) para balancear el modelo.

- Reflexionar sobre el costo en la precisión (accuracy) global al buscar un modelo más justo.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from fairlearn.metrics import MetricFrame, selection_rate, false_positive_rate, demographic_parity_difference
from fairlearn.reductions import ExponentiatedGradient, DemographicParity

# ==========================================
# 1. CONFIGURACIÓN Y GENERACIÓN DEL DATASET
# ==========================================
np.random.seed(42)
n_candidatos = 500

# Simulamos características de los solicitantes de crédito
ingresos = np.random.randint(15000, 80000, size=n_candidatos)
historial_crediticio = np.random.choice([0, 1], size=n_candidatos, p=[0.3, 0.7]) # 1 = Buen historial
genero = np.random.choice([0, 1], size=n_candidatos, p=[0.4, 0.6]) # 0 = Mujeres, 1 = Hombres

# Creamos un sesgo histórico intencional: A los hombres se les aprobaba más el crédito en el pasado
# independientemente de sus ingresos.
probabilidad_aprobacion = 0.3 * historial_crediticio + 0.4 * genero + 0.3 * (ingresos / 80000)
credito_aprobado = (probabilidad_aprobacion > 0.5).astype(int)

# Crear el DataFrame único
df = pd.DataFrame({
    'ingresos': ingresos,
    'historial': historial_crediticio,
    'genero_atributo_protegido': genero,
    'credito_aprobado_objetivo': credito_aprobado
})

# ==========================================
# 2. PREPARACIÓN DE DATOS (X, y) Y DIVISIÓN
# ==========================================
# Separamos las características (X) de la etiqueta que queremos predecir (y)
X = df[['ingresos', 'historial', 'genero_atributo_protegido']]
y = df['credito_aprobado_objetivo']

# Dividimos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Guardamos el atributo protegido de prueba por separado para Fairlearn
genero_test = X_test['genero_atributo_protegido']

print("✅ Dataset preparado y dividido correctamente.")
print(f"Tamaño de entrenamiento: {X_train.shape[0]} muestras. Tamaño de prueba: {X_test.shape[0]} muestras.\n")

# ==========================================
# 3. ENTRENAMIENTO DEL MODELO BASE (INJUSTO)
# ==========================================
modelo_base = DecisionTreeClassifier(max_depth=4, random_state=42)
modelo_base.fit(X_train, y_train)
preds_base = modelo_base.predict(X_test)

# Evaluamos el comportamiento del modelo base por género
metricas_interes = {
    'Exactitud (Accuracy)': accuracy_score,
    'Tasa de Aprobación': selection_rate,
    'Falsos Positivos (FPR)': false_positive_rate
}

marco_base = MetricFrame(
    metrics=metricas_interes,
    y_true=y_test,
    y_pred=preds_base,
    sensitive_features=genero_test
)

print("--- EVALUACIÓN DEL MODELO BASE (Por Grupo) ---")
print(marco_base.by_group)
print(f"Diferencia en la paridad demográfica: {demographic_parity_difference(y_test, preds_base, sensitive_features=genero_test):.4f}\n")

# ==========================================
# 4. MITIGACIÓN DEL SESGO (MODELO JUSTO)
# ==========================================
# Forzamos al algoritmo a aprender bajo la restricción de Paridad Demográfica
mitigador = ExponentiatedGradient(
    estimator=DecisionTreeClassifier(max_depth=4, random_state=42),
    constraints=DemographicParity()
)

# Entrenamos especificando cuál es la variable protegida en el set de entrenamiento
mitigador.fit(X_train, y_train, sensitive_features=X_train['genero_atributo_protegido'])
preds_justas = mitigador.predict(X_test)

# Evaluamos el comportamiento del nuevo modelo mitigado
marco_justo = MetricFrame(
    metrics=metricas_interes,
    y_true=y_test,
    y_pred=preds_justas,
    sensitive_features=genero_test
)

print("--- EVALUACIÓN DEL MODELO MITIGADO (Por Grupo) ---")
print(marco_justo.by_group)
print(f"Diferencia en la paridad demográfica post-mitigación: {demographic_parity_difference(y_test, preds_justas, sensitive_features=genero_test):.4f}\n")

# ==========================================
# 5. COMPARATIVA FINAL PARA EL ALUMNO
# ==========================================
print("--- COMPARATIVA GENERAL ---")
print(f"Modelo Base   -> Accuracy Global: {accuracy_score(y_test, preds_base):.2f} | Disparidad: {demographic_parity_difference(y_test, preds_base, sensitive_features=genero_test):.2f}")
print(f"Modelo Justo  -> Accuracy Global: {accuracy_score(y_test, preds_justas):.2f} | Disparidad: {demographic_parity_difference(y_test, preds_justas, sensitive_features=genero_test):.2f}")
```

El Diagnóstico del Modelo Base: Una Injusticia "Perfecta"
Al observar el primer bloque, notamos algo alarmante:

La ilusión del 100% de Exactitud: El modelo base tiene un Accuracy perfecto de 1.0 (100%). Para un alumno descuidado, este sería el "modelo ideal".

La cruda realidad del sesgo: Al desagregar los datos, la Tasa de Aprobación para el grupo 1 (Hombres) es del 96.3%, mientras que para el grupo 0 (Mujeres) es apenas del 35.2%.

**Conclusión**: El modelo es "perfectamente exacto" únicamente porque aprendió a replicar a la perfección los prejuicios históricos del pasado. Logró la máxima eficiencia técnica a costa de una enorme injusticia social (una disparidad del 0.61).
