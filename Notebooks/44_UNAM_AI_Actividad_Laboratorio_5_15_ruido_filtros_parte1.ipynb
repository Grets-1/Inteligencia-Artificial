```python
import matplotlib.pyplot as plt
import cv2 as cv
import numpy as np
```

```python
def mostrar_imagen(img1, img2, titulo1 = 'Original', titulo2 = 'Nueva'):
  plt.figure(figsize=(10,10))
  plt.subplot (1,2,1)
  plt.imshow(cv.cvtColor(img1, cv.COLOR_BGR2RGB))
  plt.title(titulo1)
  plt.subplot(1,2,2)
  plt.imshow(cv.cvtColor(img2, cv.COLOR_BGR2RGB))
  plt.title(titulo2)
  plt.show()
```

```python
img = cv.imread('220px-Dante_Alighieri01.jpg')
plt.imshow(cv.cvtColor(img,cv.COLOR_BGR2RGB))
```

```python
renglones, cols, _= img.shape
ruido = np.random.normal(0,15,(renglones, cols, 3)).astype(np.uint8)
ruido
```

```python
img_ruido = img + ruido
mostrar_imagen(img, img_ruido,titulo1='Original', titulo2 = 'Imagen con Ruido')
```

```python
renglones, cols, _ = img.shape
ruido = np.random.normal(0,50,(renglones, cols, 3)).astype(np.uint8)
ruido
```

```python

img_ruido = img + ruido
mostrar_imagen(img, img_ruido,titulo1='Original', titulo2 = 'Imagen con Ruido de 50')
```

```python
kernel = np.ones((6,6))/25
kernel
```

```python
img_filtrado = cv.filter2D(src = img_ruido, ddepth = -1, kernel = kernel)
mostrar_imagen(img_ruido,img_filtrado,titulo2='Imagen Filtrada', titulo1 = 'Imagen con Ruido de 50')
```

```python
kernel = np.ones((4,4))/16
kernel
```

```python
img_filtrado = cv.filter2D(src = img_ruido, ddepth = -1, kernel = kernel)
mostrar_imagen(img_ruido,img_filtrado,  titulo2 = 'Imagen con Ruido de 50',titulo1='Imagen Filtrada')
```

```python
img_filtrada = cv.GaussianBlur(img_ruido,(5,5),sigmaX=4, sigmaY=4)
mostrar_imagen( img_ruido,img_filtrada,titulo2 = 'Imagen con Ruido de Gaussina Blur sigma 4',titulo1='Imagen Ruido')
```

```python
img_filtrada = cv.GaussianBlur(img_ruido,(5,5),sigmaX=10, sigmaY=10)
mostrar_imagen( img_ruido,img_filtrada,titulo2 = 'Imagen con GaussianBlur sigma 10',titulo1='Imagen Filtrada')
```

```python
kernel = np.array([[-1,-1,-1],
                   [-1,9,-1],
                   [-1,-1,-1]])
img_nitida = cv.filter2D(img, -1, kernel)
mostrar_imagen( img,img_nitida,titulo1='Imagen Original',titulo2="Imagen Nítida")
```

```python
img_grises= cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
img_grises =cv.GaussianBlur(img_grises,(3,3),sigmaX=0.1,sigmaY=0.1)
ddepth = cv.CV_16S
grad_x= cv.Sobel(src=img_grises,ddepth=ddepth,dx=1,dy=0,ksize=3)
grad_y= cv.Sobel(src=img_grises,ddepth=ddepth,dx=0,dy=1,ksize=3)

abs_grad_x= cv.convertScaleAbs(grad_x)
abs_grad_y= cv.convertScaleAbs(grad_y)

grad = cv.addWeighted(abs_grad_x, 0.5, abs_grad_y,0.5,0)
mostrar_imagen(img_grises, grad, titulo1="Imagen Original",titulo2='Deteccion de bordes')
```

```python
img_grises= cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
img_filtrada =cv.medianBlur(img_grises,5)
mostrar_imagen(img_grises,img_filtrada, titulo1='Imagen original', titulo2='Imagen filtrada')
```

```python
img_grises = cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
kernel = np.ones((4,4),np.uint8)
kernel
```

```python

img_erosion = cv.erode(img_grises, kernel,iterations = 1)
mostrar_imagen(img_grises,img_erosion, titulo1='Imagen original', titulo2='Imagen erosionada')
```

```python
img_grises = cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
kernel = np.ones((4,4),np.uint8)
kernel
```

```python

img_dilatada = cv.dilate(img_grises, kernel,iterations = 1)
mostrar_imagen(img_grises,img_dilatada, titulo1='Imagen original', titulo2='Imagen dilatada')
```

```python
img_grises = cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
kernel = np.ones((4,4),np.uint8)
kernel
```

```python

img_erosion = cv.morphologyEx(img_grises, cv.MORPH_OPEN, kernel)
mostrar_imagen(img_grises,img_erosion, titulo1='Imagen original', titulo2='Imagen con apertura')
```

```python
img_grises = cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
kernel = np.ones((4,4),np.uint8)
img_erosion = cv.morphologyEx(img_grises, cv.MORPH_CLOSE, kernel)
mostrar_imagen(img_grises,img_erosion, titulo1='Imagen original', titulo2='Imagen con cierre')
```

```python
img_grises = cv.imread('220px-Dante_Alighieri01.jpg',cv.IMREAD_GRAYSCALE)
kernel = np.ones((4,4),np.uint8)
kernel
```

```python
img_erosion = cv.morphologyEx(img_grises, cv.MORPH_GRADIENT, kernel)
mostrar_imagen(img_grises,img_erosion, titulo1='Imagen original', titulo2='Gradiante de la imagen')
```

# Conclusión

El presente laboratorio exploró diversas técnicas fundamentales en el procesamiento de imágenes utilizando OpenCV y Python. Se comenzó por la introducción de ruido gaussiano en una imagen, lo cual simuló condiciones reales de degradación visual. Posteriormente, se aplicaron diferentes métodos de filtrado para reducir este ruido, incluyendo filtros promediadores (con kernels de distintos tamaños) y el filtro Gaussiano, observando cómo estos suavizan la imagen en diferentes grados.

Adicionalmente, se demostró la capacidad de realzar detalles de una imagen mediante un filtro de nitidez. Se exploró la detección de bordes utilizando el operador de Sobel, una técnica crucial para identificar contornos y estructuras importantes en la imagen.

Finalmente, el laboratorio profundizó en las operaciones morfológicas básicas: erosión y dilatación. Estas operaciones, junto con sus combinaciones como la apertura y el cierre, fueron utilizadas para limpiar imágenes, eliminar ruido pequeño y rellenar orificios, mostrando su utilidad en el análisis de formas y la segmentación. En conjunto, estos ejercicios proporcionaron una comprensión práctica de cómo diversas transformaciones pueden ser aplicadas a las imágenes para su mejora, análisis y preparación para tareas de visión por computadora más complejas.
